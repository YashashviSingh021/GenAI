```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite", temperature=0)
out = StrOutputParser()
```

```python
####Transform Case to make the output in upper case

def transform_case(result: str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator and translate the input in {language}."),
    ("human", "{query}"),
])
```

```python
### Insted of this we can use chain to create a prompt and then pass it to the llm.invoke method.
# final_prompt = prompts.format_prompt(language="Hindi", query="Hello, how are you?").to_messages()
# res= llm.invoke(final_prompt)
# print(res.content)
```

```python
#####Chains 
chains = prompts | llm | out | transform_case
response = chains.invoke({
    "language": "Hinglish",
    "query": "Hello, how are you?"
})
print(response)
```
