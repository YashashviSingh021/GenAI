```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
###Here we are creating tool which will add two numbers and return the result. This tool can be used in LangChain agents to perform addition of two numbers and we also used the tool decorator to make it a tool which can be used in LangChain agents.
@tool
def add_two_numbers(a:int, b:int):
    """
    it will add the two numbers and return the result

    Args:
        a : Number one
        b: Number two
    """
    return a + b

@tool
def multiply_two_numbers(a:int, b:int):
    """
    it will multiply the two numbers and return the result

    Args:
        a : Number one
        b: Number two
    """
    return a * b
```

```python
add_two_numbers.invoke({"a":2, "b":3})
```

```python
###Here we are creating a LangChain agent which will use the above two tools to perform addition and multiplication of two numbers. We are also providing a system prompt to the agent
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
###Here we are creaeting the agent and to create the agent we are passing the model, tools and system prompt to the create_agent function. The system prompt is used to provide the context to the agent about what it is supposed to do. In this case we are telling the agent that it is a math teacher and it will use the tools for calculation.
agent = create_agent(
    model = llm,
    tools = [add_two_numbers, multiply_two_numbers],
    system_prompt = "You are a math teacher and you will use the tools for calculation"
)
```

```python
###Here we are invoking the agent and you can see that it is fixed that we have to pass messages in the form of a list of dictionaries where each dictionary will have two keys role and content. The role can be user or system and the content is the message which we want to pass to the agent. In this case we are passing a message from the user asking the agent to calculate the sum of 10 + 10.
response = agent.invoke({"messages": [{"role": "user", "content": "what is the sum of 10 + 10"}]})
print(response["messages"][-1].content)
```
