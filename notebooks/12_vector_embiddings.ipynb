```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "machine learning is a subset of AI",
    "deep learning usees neural networks",
    "python is a programming lang",
    "india won cricket worldcup in 2011"
]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model="gemini/embedding-001")
```

```python
vector = embeddings.aembed_documents(documents)
```

```python
len(vector)
```

Store Vector Embidding into vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts = documents,
    embedding = embeddings,
    persist_directory = "./vector_db"
)
```

```python
query = "machine Learning"
result = vector_store.similarity_search(query = query)
result
```
