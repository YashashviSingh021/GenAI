# Notebook Summary

- Loaded environment variables using `python-dotenv`.
- Initialized a Google Gemini language model with LangChain.
- Extracted a user’s name, email, and age from text.
- Used Pydantic models for structured output validation.
- Generated a structured list of movies with titles and release years.

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import List
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
text = "Hello, my name is Yash Singh,"\
"Email is yash.singh@example.com and my age is 25"
res = llm.invoke(f"Please give me only name, email and age from this text: {text}")
print(res.content)
```

Now Structured Output

```python
from pydantic import BaseModel, Field

class ResponseStructure(BaseModel):
    name: str = Field(description="Name of the User")
    email: str = Field(description="Email of the User")
    age: int = Field(description="Age of the User")

structured_llm = llm.with_structured_output(ResponseStructure)
###here we use f because we want to pass the text variable into the prompt
res = structured_llm.invoke(f"Please give me only name, email and age from this text: {text}")
res
```

Now Structured OutPut for mutiple values

```python
class Movies(BaseModel):
    title: str = Field(description="Title of the Movie")
    year: int = Field(description="Year of the Movie")

class AllMovies(BaseModel):
    movies: List[Movies]
movies_llm = llm.with_structured_output(AllMovies)
```

```python
res= movies_llm.invoke("Please give me 4 movies with their names and years of release")
res
```
