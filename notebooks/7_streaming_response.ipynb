```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite", streaming=True)
```

```python
question = "Please explain me the Dependence Injection in .Net with an example in C# and also provide me the code for the same."
```

```python
response = llm.stream(question)
###here we are using isinstance to check the type of content in the response and print it accordingly. If the content is a string, we print it directly. If it's a list, we iterate through the list and print the text from any dictionary items that have a "type" of "text".

### Earlier we are using response.content[0]["text"] to print the content but now using isinstance to check the type of content in the response and print it accordingly. If the content is a string, we print it directly. If it's a list, we iterate through the list and print the text from any dictionary items that have a "type" of "text".
for chunk in response:
    if isinstance(chunk.content, str):
        print(chunk.content, end="")

    elif isinstance(chunk.content, list):
        for content in chunk.content:
            if isinstance(content, dict) and content.get("type") == "text":
                print(content.get("text", ""), end="")
```
